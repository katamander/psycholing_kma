```python
# import
import pandas as pd
```

```python
# load csv
df = pd.read_csv("../materials/sequences.csv")

# calc character length
df["G_len"] = df["grammatical"].str.len()
df["U_len"] = df["ungrammatical"].str.len()
```

```python
# print summary
print(df[["G_len", "U_len"]].describe().round(2))
```

```python
# plot distribution
ax = df[["G_len", "U_len"]].plot.hist(bins = 15, alpha = 0.5)
ax.set_xlabel("Length (characters)")
ax.set_ylabel("Count")
```
