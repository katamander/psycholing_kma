```python
# import
import pandas as pd
import seaborn as sns
```

```python
# load raw data
raw = pd.read_csv("/work/E2026/psycholing_kma/data/combined_data.csv")
raw.head(20)
```

```python
# get list of column names for ease
list(raw)
```

```python
# make list of relevant columns
cols = ['participant_id', 'sender', 'duration', 'prime', 'target', 'response', 'condition', 'soa_condition', 'congruence', 'correct_response', 'correct']
cols
```

```python
# make new df with only relevant columns
df = raw[cols]
df
```

```python
# keep only target rows i.e. filter out welcome screens etc
df = df[df['sender'] == 'target']
# df['sender'].unique()
# filter out practice rows
df = df[df['condition'] != 'practice']
# drop now-irrelevant sender column
df = df.drop(columns = ['sender'])
df
```

```python
# correct error
df.loc[df['target'] == 'BULL', 'correct_response'] = 'n'
#df['accuracy'] = df['response'] == df['correct_response']
df
```

```python
df_grouped = df.groupby(['soa_condition', 'congruence']).mean(['duration'])
df_grouped
```

```python
df_grouped = df_grouped.reset_index()
df_grouped
```

```python
ax = sns.barplot(data = df_grouped, x = 'soa_condition', y = 'duration', hue = 'congruence')
# sns.move_legend(ax, "lower right")
```
