### data scrambler

```python
# import
import numpy as np
import os
import pandas as pd
```

```python
# define scramble function
def scramble_data(in_file, out_file):
    # load semicolon csv as df with row 1 as header cos that's how surveyxact works
    df = pd.read_csv(os.path.join("..", "in", in_file), sep = ";", header = 1, encoding = "utf-8")
    #scramble!
    for col in df.columns:
        # anonymise open-text comments in the comment column
        if "kommentarer" in col.lower():
            df.loc[df[col].notna(), col] = "Denne kommentar er anonymiseret."
        # shuffle all other columns randomly
        else:
            df[col] = np.random.permutation(df[col].values)
    # save to csv
    with open(os.path.join("..", "in", out_file), "w", encoding = "utf-8") as f:
        f.write("scrambled survey data\n") # marks as scrambled data & preserves header = 1 from original
        df.to_csv(f, sep = ";", index = False)
    # print statement
    print("scrambled dataset saved to /in")
```

```python
# apply scramble function
scramble_data("original_survey_data.csv", "survey_data.csv")
```
